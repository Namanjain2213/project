```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.linspace(0,5,11)
y = x ** 2 
```

```python
plt.plot(x,y)
plt.xlabel('value')
plt.ylabel('square value')
plt.title('trend of square value')
```

```python
plt.subplot(1,2,1)
plt.plot(x,y)
plt.subplot(1,2,2)
plt.plot(y,x)
plt.subplot(1,2)
```

```python
plt.scatter(x,y)
```

```python
plt.boxplot(x)
```

```python
plt.plot(x,y,color='red')
```
